```python
from pyspark.sql import functions as F
```

##Inspect Delta tables

```python
tables = [
    "workspace.bronze.transactions",
    "workspace.silver.transactions",
    "workspace.silver.transaction_quarantine",
    "workspace.gold.account_balances",
    "workspace.gold.fraud_alerts",
    "workspace.gold.daily_transaction_metrics",
    "workspace.gold.data_quality_results"
]

for table in tables:
    print(table)
    display(
        spark.sql(f"DESCRIBE DETAIL {table}")
    )
```

##Enable Delta optimization

```python
optimization_tables = [
    "workspace.bronze.transactions",
    "workspace.silver.transactions",
    "workspace.silver.transaction_quarantine",
    "workspace.gold.account_balances",
    "workspace.gold.fraud_alerts",
    "workspace.gold.daily_transaction_metrics"
]

for table in optimization_tables:
    spark.sql(f"""
        ALTER TABLE {table}
        SET TBLPROPERTIES (
            'delta.autoOptimize.optimizeWrite' = 'true',
            'delta.autoOptimize.autoCompact' = 'true'
        )
    """)

print("Delta optimization properties configured")
```

##Optimize Silver

```python
%sql
OPTIMIZE workspace.silver.transactions
ZORDER BY (account_id, txn_time);
```

##Optimize Gold

##Test Time Travel safely

```python
silver_history = spark.sql("""
    DESCRIBE HISTORY workspace.silver.transactions
""")

versions = (
    silver_history
    .select("version")
    .orderBy(F.desc("version"))
    .collect()
)

print("Available versions:")

for row in versions:
    print(row["version"])
```

##Read an older version

```python
old_silver = spark.sql("""
    SELECT *
    FROM workspace.silver.transactions VERSION AS OF 0
""")

print(
    "Historical Silver records:",
    f"{old_silver.count():,}"
)
```
